# 01 — What does a decoder actually see?

A surface-code decoder never sees qubits, or errors, or a code. It sees a list of
bits: which stabilizer measurements disagreed with the round before. This notebook
builds that object from the code up, then asks what a standard matching decoder
does and does not read from it.

The answer is the reason the rest of this project exists:

> The standard *graphlike* error model splits the detection events into two halves
> that share no edges at all. The half a logical-$Z$ memory experiment supposedly
> does not need turns out to predict, well above chance, when matching gets the
> answer wrong.

Nothing here is a decoder, and nothing here involves hardware. Both come next.

**Outputs:** `figures/code_geometry.png`, `figures/detector_graph.png`,
`figures/discarded_information.png`, `results/detector_model_facts.csv`.

## 1. The code

`RotatedSurfaceCode` is built from the stabilizer rule alone in
[`qec_utils.py`](qec_utils.py), next to this notebook. The cell below is where it
gets checked: if the logical operators were not minimum
weight, or the two check families did not commute, nothing after this point would
mean anything.

$d^2$ data qubits, $d^2 - 1$ independent stabilizers split evenly between $X$ and $Z$
type, one encoded logical qubit. $H_X$ and $H_Z$ have one row per check and one
column per data qubit, and CSS commutation is $H_X H_Z^{\mathsf T} = 0 \pmod 2$.

```python
codes = {d: RotatedSurfaceCode(d) for d in (3, 5)}

for d, code in codes.items():
    lx = code.vector(code.logical_x)
    lz = code.vector(code.logical_z)
    assert code.n_data == d * d
    assert code.n_stabilizers == d * d - 1
    assert len(code.x_checks) == len(code.z_checks) == (d * d - 1) // 2
    assert code.n_logical == 1
    assert not np.any((code.hx @ code.hz.T) % 2)    # CSS commutation
    assert not np.any((code.hx @ lz) % 2)           # Z-bar commutes with every X check
    assert not np.any((code.hz @ lx) % 2)           # X-bar commutes with every Z check
    assert np.dot(lx, lz) % 2 == 1                  # the two logicals anticommute
    assert len(code.logical_x) == len(code.logical_z) == d
    # ...and neither is a product of stabilizers, which is what makes it logical
    # rather than trivial: a vector is in the row space only if adding it to the
    # matrix leaves the rank alone.
    assert gf2_rank(np.vstack([code.hx, lx])) > gf2_rank(code.hx)
    assert gf2_rank(np.vstack([code.hz, lz])) > gf2_rank(code.hz)
    print(f"d = {d}:  {code.n_data} data, {len(code.x_checks)}X + "
          f"{len(code.z_checks)}Z checks, k = {code.n_logical}, "
          f"logical weight {len(code.logical_x)}  OK")
```

## 2. The memory circuit

A memory experiment does nothing except try to keep a logical state alive. Prepare
the logical $|0\rangle$, measure every stabilizer $d$ times, then measure all the
data qubits. If the code works, the logical value read out at the end matches what
went in.

Stim's `surface_code:rotated_memory_z` is the reference schedule. Its structure is
worth reading once: a reset, then $d$ identical rounds of Hadamards, four layers of
`CX`, and a measure-and-reset of every ancilla, then a final data measurement.

```python
def memory_circuit(d, p=0.0, rounds=None):
    # Uniform circuit-level noise. p = 0 gives the ideal circuit.
    return stim.Circuit.generated(
        "surface_code:rotated_memory_z", distance=d, rounds=rounds or d,
        after_clifford_depolarization=p, before_measure_flip_probability=p,
        after_reset_flip_probability=p, before_round_data_depolarization=p,
    )

summary = []
for d in (3, 5):
    ideal = memory_circuit(d)
    summary.append({
        "d": d, "rounds": d,
        "qubits": ideal.num_qubits,
        "measurements": ideal.num_measurements,
        "detectors": ideal.num_detectors,
        "observables": ideal.num_observables,
    })
pd.DataFrame(summary).set_index("d")
```

```python
# The first round, verbatim. Ancillas 2, 11, 16, 25 get Hadamards: those are the X
# checks. The four CX layers are the schedule that keeps hook errors harmless.
print("\n".join(str(memory_circuit(3)).splitlines()[17:42]))
```

## 3. Detection events

A **detector** is a parity of measurement outcomes that is *deterministic* when
nothing goes wrong -- almost always "this ancilla now, XOR the same ancilla last
round". A detector fires when that parity comes out wrong. So the decoder's input
is not a syndrome in the textbook sense; it is a list of *changes*.

Two consequences follow immediately, and they are why the decoder's graph lives in
space **and** time:

* a data-qubit error fires two detectors in the **same round**, at the two checks
  that straddle it -- a spatial pair;
* a measurement error fires the **same detector in two consecutive rounds**, because
  the misread propagates into the next round's comparison -- a temporal pair.

With no noise, nothing fires at all.

```python
det, obs = memory_circuit(3).compile_detector_sampler().sample(
    1000, separate_observables=True)
assert det.sum() == 0 and obs.sum() == 0
print(f"noiseless: {det.sum()} detection events and {obs.sum()} observable flips "
      f"across {det.shape[0]} shots x {det.shape[1]} detectors  OK")
```

```python
# Inject one deterministic fault into the ideal circuit and see exactly what fires.
# Flattening first removes REPEAT blocks so a line index means what it looks like.
ideal_flat = memory_circuit(3).flattened()
lines = str(ideal_flat).splitlines()
coords3 = ideal_flat.get_detector_coordinates()

def sector(i, coords=coords3):
    # X ancillas sit where (x + y) / 2 is odd; Z ancillas where it is even.
    x, y, _ = coords[i]
    return "X" if ((int(x) + int(y)) // 2) % 2 == 1 else "Z"

def label(i, coords=coords3):
    x, y, t = (int(v) for v in coords[i])
    return f"({x},{y})@t{t}[{sector(i, coords)}]"

def inject(instruction, after_line):
    modified = list(lines)
    modified.insert(after_line, instruction)
    d, o = stim.Circuit("\n".join(modified)).compile_detector_sampler().sample(
        1, separate_observables=True)
    return np.flatnonzero(d[0]), int(o[0, 0])

# Inject after round 1's ancilla measurement, where both Pauli components are live.
after_round_1 = next(i for i, l in enumerate(lines) if l.strip().startswith("MR ")) + 1

faults = [
    ("X on data (1,1)",          "X_ERROR(1) 1",  after_round_1),
    ("Z on data (1,1)",          "Z_ERROR(1) 1",  after_round_1),
    ("Y on data (1,1)",          "Y_ERROR(1) 1",  after_round_1),
    ("X on data (3,3), bulk",    "X_ERROR(1) 10", after_round_1),
    ("Y on data (3,3), bulk",    "Y_ERROR(1) 10", after_round_1),
    ("misread of ancilla (2,2)", "X_ERROR(1) 9",  after_round_1 - 1),
]
rows = []
for name, instruction, where in faults:
    fired, flip = inject(instruction, where)
    rows.append({
        "fault": name,
        "detectors fired": " ".join(label(i) for i in fired) or "none",
        "sectors": "+".join(sorted({sector(i) for i in fired})) or "-",
        "flips observable": flip,
    })
pd.DataFrame(rows).set_index("fault")
```

Read the last column against the third.

An $X$ error on the corner qubit $(1,1)$ fires **one** $Z$-sector detector and flips
the logical observable. A $Y$ error on the same qubit fires **that same $Z$-sector
detector, plus an $X$-sector one**, and also flips the observable. The misread
ancilla fires the same detector twice in a row and flips nothing.

So the $X$-sector detectors are not noise. They carry information about which
physical fault happened, and therefore about whether the logical frame moved. The
question for the rest of this notebook is whether a matching decoder uses it.

## 4. The detector graph, in space and time

Stim compiles a noisy circuit into a **detector error model**: a list of independent
fault mechanisms, each with a probability, the detectors it fires, and whether it
flips the observable. Asking for `decompose_errors=True` additionally forces every
mechanism into pieces touching at most two detectors, because that is what a
matching decoder can consume -- pieces of two become graph edges, pieces of one
become edges to a virtual boundary.

That decomposition is where the trouble starts.

```python
def mechanisms(dem):
    # Each DEM error becomes (probability, [component, ...], observable_flip), where
    # a component is the detector list between "^" separators. Without decomposition
    # there is exactly one component, which may touch any number of detectors.
    out = []
    for instruction in dem.flattened():
        if instruction.type != "error":
            continue
        components, current, flip = [], [], 0
        for target in instruction.targets_copy():
            if target.is_separator():
                components.append(current); current = []
            elif target.is_relative_detector_id():
                current.append(target.val)
            elif target.is_logical_observable_id():
                flip ^= 1
        components.append(current)
        out.append((instruction.args_copy()[0], components, flip))
    return out

noisy3 = memory_circuit(3, P)
dem_graphlike = noisy3.detector_error_model(decompose_errors=True)
dem_true = noisy3.detector_error_model(decompose_errors=False)

graph = nx.Graph()
graph.add_nodes_from(range(dem_graphlike.num_detectors))
for _, comps, _ in mechanisms(dem_graphlike):
    for comp in comps:
        if len(comp) == 2:
            graph.add_edge(*comp)

parts = sorted(nx.connected_components(graph), key=len, reverse=True)
print(f"graphlike detector graph: {graph.number_of_nodes()} detectors, "
      f"{graph.number_of_edges()} edges")
print(f"connected components: {len(parts)}")
for part in parts:
    print(f"   {len(part):2d} detectors, sector(s) {sorted({sector(i) for i in part})}")

# Not a figure of speech: the two check families share no edge whatsoever.
assert len(parts) == 2
assert all(len({sector(i) for i in part}) == 1 for part in parts)
```

```python
purple = "tab:purple"
colour = {"X": blue, "Z": orange}
position = {i: tuple(int(v) for v in dem_graphlike.get_detector_coordinates()[i])
            for i in range(dem_graphlike.num_detectors)}

edges = {(min(a, b), max(a, b)) for _, comps, _ in mechanisms(dem_graphlike)
         for comp in comps if len(comp) == 2 for a, b in [comp]}
cross_sector = [(p, c[0]) for p, c, _ in mechanisms(dem_true)
                if len({sector(i) for i in c[0]}) > 1]
featured = sorted(cross_sector, key=lambda t: -t[0])[:12]

figure = plt.figure(figsize=(11.5, 4.8))
for panel, show_cross in enumerate((False, True)):
    ax = figure.add_subplot(1, 2, panel + 1, projection="3d")
    for a, b in edges:
        xs, ys, ts = zip(*(position[a], position[b]))
        ax.plot(xs, ys, ts, color=colour[sector(a)], lw=1.1,
                alpha=0.25 if show_cross else 0.55)
    if show_cross:
        for _, dets in featured:
            centre = np.mean([position[i] for i in dets], axis=0)
            for i in dets:
                ax.plot([centre[0], position[i][0]], [centre[1], position[i][1]],
                        [centre[2], position[i][2]], color=purple, lw=1.7, alpha=0.85)
    for which in ("Z", "X"):
        pts = np.array([position[i] for i in position if sector(i) == which])
        ax.scatter(pts[:, 0], pts[:, 1], pts[:, 2], s=30, c=colour[which],
                   edgecolors="white", linewidths=0.7, depthshade=False)
    ax.set_xlabel("x", fontsize=8, labelpad=-8)
    ax.set_ylabel("y", fontsize=8, labelpad=-8)
    ax.set_zlabel("round", fontsize=8, labelpad=-8)
    ax.set_zticks(range(4))
    ax.tick_params(labelsize=6.5, pad=-3, colors=gray)
    ax.set_box_aspect((1, 1, 0.85))
    ax.view_init(elev=17, azim=-62)
    title, subtitle = [
        ("What MWPM is given", "graphlike model: two components, no edge between them"),
        ("What the noise actually does",
         f"12 likeliest of {len(cross_sector)} cross-sector mechanisms"),
    ][panel]
    ax.set_title(title, fontsize=11.5, pad=-4, weight="bold")
    ax.text2D(0.5, 0.95, subtitle, transform=ax.transAxes, ha="center",
              fontsize=8.5, color=gray)

handles = [
    plt.Line2D([], [], marker="o", ls="none", color=colour["Z"], label="$Z$-check detector"),
    plt.Line2D([], [], marker="o", ls="none", color=colour["X"], label="$X$-check detector"),
    plt.Line2D([], [], color=purple, lw=2, label="one fault firing both sectors"),
]
figure.legend(handles=handles, loc="lower center", ncol=3, frameon=False, fontsize=9,
              bbox_to_anchor=(0.5, -0.02), labelcolor=gray)
figure.subplots_adjust(left=0.01, right=0.99, top=0.99, bottom=0.06, wspace=0.02)
figure.savefig("figures/detector_graph.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 5. What decomposition costs

The left panel is not a stylised diagram. The graph handed to a matching decoder
genuinely has two connected components: the $Z$-check detectors and the $X$-check
detectors, with no edge anywhere between them.

The right panel shows what the circuit's own noise model says. Every violet star is
one fault -- a $Y$ error, or a two-qubit depolarizing event on a `CX` -- that fires
detectors in *both* families at once. Decomposition cannot represent those, so it
splits each into an $X$-sector piece and a $Z$-sector piece and lets the decoder
treat them as independent events.

How much gets split this way?

```python
facts = []
for d in (3, 5):
    circuit = memory_circuit(d, P)
    true = mechanisms(circuit.detector_error_model(decompose_errors=False))
    graphlike = mechanisms(circuit.detector_error_model(decompose_errors=True))

    sizes = pd.Series([len(c[0]) for _, c, _ in true]).value_counts().sort_index()
    total_mass = sum(p for p, _, _ in true)
    coords_d = circuit.detector_error_model().get_detector_coordinates()
    cross = [(p, c[0]) for p, c, _ in true
             if len({sector(i, coords_d) for i in c[0]}) > 1]
    facts.append({
        "d": d,
        "fault mechanisms": len(true),
        "after decomposition": len(graphlike),
        "hyperedges (>2 detectors)": sum(1 for _, c, _ in true if len(c[0]) > 2),
        "cross-sector mechanisms": len(cross),
        "cross-sector share of error probability":
            f"{sum(p for p, _ in cross) / total_mass:.1%}",
    })
    if d == 3:
        print("detectors per fault mechanism, before decomposition:")
        for size, count in sizes.items():
            print(f"   {size} detector(s): {count:4d} mechanisms")
        print()

# Every decomposed component is small enough to be a graph edge; that is the point.
assert max(len(c) for _, comps, _ in graphlike for c in comps) <= 2
pd.DataFrame(facts).set_index("d")
```

## 6. What a matching decoder actually reads

Roughly a fifth of the error probability sits in mechanisms that link the two
sectors, and decomposition severs every one of them.

That is an argument about the model. Here is the corresponding fact about the
decoder. Take real shots, decode them with PyMatching, then decode *the same shots
again* with every $X$-sector detector bit erased, and compare the predictions.

If the $X$ sector mattered to the decoder, erasing it would change something.

```python
N_SHOTS = 200_000
blinding, samples = [], {}

for d in (3, 5):
    circuit = memory_circuit(d, P)
    dem = circuit.detector_error_model(decompose_errors=True)
    coords_d = dem.get_detector_coordinates()
    is_x = np.array([sector(i, coords_d) == "X" for i in range(dem.num_detectors)])

    detections, observable = circuit.compile_detector_sampler(seed=2026).sample(
        N_SHOTS, separate_observables=True)
    matcher = pymatching.Matching.from_detector_error_model(dem)
    predicted = matcher.decode_batch(detections)[:, 0]

    blinded = detections.copy()
    blinded[:, is_x] = False
    predicted_blind = matcher.decode_batch(blinded)[:, 0]

    changed = int((predicted != predicted_blind).sum())
    samples[d] = (detections, observable[:, 0], predicted, is_x)
    blinding.append({
        "d": d,
        "shots": N_SHOTS,
        "shots where an X detector fired": f"{detections[:, is_x].any(axis=1).mean():.1%}",
        "MWPM logical failure": f"{(predicted != observable[:, 0]).mean():.2e}",
        "no correction": f"{observable[:, 0].mean():.2e}",
        "predictions changed by erasing X sector": changed,
    })
    # The X sector is not merely under-used. It is unread.
    assert changed == 0

pd.DataFrame(blinding).set_index("d")
```

At $d = 5$ an $X$-sector detector fires in more than nine shots out of ten, and
erasing all of them changes **zero** of 200,000 predictions.

This is not a flaw in PyMatching. It is correct behaviour given its input: in the
graphlike model no $X$-sector edge carries the logical observable, so no amount of
$X$-sector evidence can move the answer. The information was discarded upstream, at
the moment the error model was decomposed.

## 7. Is the discarded information worth anything?

"Unread" is not the same as "useless". The honest test is whether those bits predict
something the decoder cannot already work out.

So: take the shots, ask whether MWPM got each one **wrong**, and try to predict that
from the $X$-sector bits alone using nothing more than logistic regression. Fit on
half the shots, score the other half. A shuffled copy of the same features is the
control -- it destroys the correspondence between a shot's $X$ bits and its outcome
while keeping every other statistical property.

If the control lands at 0.5 and the real features land well above it, the $X$ sector
carries decision-relevant information that matching threw away.

```python
auc, curves = [], {}
rng = np.random.default_rng(0)

for d, (detections, truth, predicted, is_x) in samples.items():
    residual = (predicted ^ truth).astype(int)     # 1 exactly where MWPM was wrong
    half = len(residual) // 2
    row = {"d": d, "MWPM failures in test half": int(residual[half:].sum())}
    for name, columns, shuffle in [
        ("X-sector bits", is_x, False),
        ("Z-sector bits", ~is_x, False),
        ("X-sector, shuffled (control)", is_x, True),
    ]:
        features = detections[:, columns].astype(np.float32)
        if shuffle:
            features = features.copy()
            rng.shuffle(features)
        model = LogisticRegression(max_iter=2000, class_weight="balanced")
        model.fit(features[:half], residual[:half])
        score = model.decision_function(features[half:])
        row[name] = roc_auc_score(residual[half:], score)
        if d == 5:
            curves[name] = roc_curve(residual[half:], score)
    auc.append(row)

table = pd.DataFrame(auc).set_index("d")
assert (table["X-sector bits"] > 0.6).all()
assert table["X-sector, shuffled (control)"].between(0.45, 0.55).all()
table
```

```python
figure, ax = plt.subplots(figsize=(5.8, 5.4))
styling = {
    "Z-sector bits": (orange, "-", 2.0),
    "X-sector bits": (blue, "-", 2.0),
    "X-sector, shuffled (control)": (gray, "--", 1.6),
}
ax.plot([0, 1], [0, 1], color="lightgray", lw=1.2, zorder=0)
for name, (fpr, tpr, _) in curves.items():
    line_colour, style, width = styling[name]
    # ROC curves all converge at (0,0) and (1,1), so a floating direct label
    # collides with a neighbour at any offset. The value sits beside its swatch.
    ax.plot(fpr, tpr, color=line_colour, ls=style, lw=width, zorder=2,
            label=f"{name}  ·  AUC {table.loc[5, name]:.3f}")
ax.set_xlabel("false positive rate", fontsize=10)
ax.set_ylabel("true positive rate", fontsize=10)
ax.set_title("Predicting MWPM's mistakes at $d = 5$,\nfrom detector bits it never reads",
             fontsize=12.5, pad=12)
ax.set_xlim(0, 1); ax.set_ylim(0, 1.001); ax.set_aspect("equal")
ax.tick_params(labelsize=8, colors=gray)
for spine in ("top", "right"):
    ax.spines[spine].set_visible(False)
ax.legend(loc="lower right", frameon=False, fontsize=9.5, labelcolor=gray)
figure.tight_layout()
figure.savefig("figures/discarded_information.png", dpi=200, bbox_inches="tight")
plt.show()
```

## 8. Where this goes

Four facts, each measured above rather than asserted:

1. **19% of the error probability** at $d = 3$ and **24%** at $d = 5$ sits in fault
   mechanisms that fire both check families at once.
2. Graphlike decomposition splits every one of them, leaving a decoder graph with
   **two connected components**.
3. A matching decoder therefore **never reads** the $X$ sector: erasing it changed
   0 of 200,000 predictions, in shots where those detectors fired over 90% of the
   time.
4. Those same bits predict **when matching is wrong** at AUC around 0.7, against 0.5
   for a shuffled control -- from logistic regression on raw bits, with no model of
   the code at all.

Point 4 is a lower bound, produced by the least capable model available. That is the
opening the rest of the project aims at.

**Notebook 02** replaces the uniform error rate `P` used throughout this notebook
with per-qubit rates from a frozen IBM calibration snapshot, and establishes the
MWPM baseline under matched controls. It also states plainly why a rotated surface
code cannot be laid out on that hardware without routing -- the heavy-hex lattice
has degree 3 and these ancillas need degree 4 -- so the snapshot is used as a source
of *realistic error rates*, not as a claim about running this circuit on that machine.

**Notebook 03** trains one small decoder that sees both sectors, and reports whether
it converts the gap measured here into a lower logical failure rate.

One caveat to carry forward: correlated decoding is a known idea, and the gains
reported in the literature are real but modest, on the order of a 10% improvement in
effective distance. Nothing above predicts a large win. It predicts a *measurable*
one, and the job of notebook 03 is to measure it honestly either way.
